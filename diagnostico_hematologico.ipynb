```python
from experta import Fact
from Expertaaa import get_diagnosis
```

```python
def SioNo(answer):
    while True:
        answer = answer.strip().lower()
        if answer == 's':
            return answer
        elif answer == 'n':
            return answer
        else:
            answer = input("Por favor, ingresa 's' para Sí o 'n' para No: ")

def show_symptoms_menu():
    print("SÍNTOMAS DISPONIBLES:")
    print("0. No tengo ninguno de estos síntomas")
    symptoms = [
        "fatigue", "paleness", "fragile nails", "yellow skin", "sore tongue",
        "fever", "bleeding", "bruises", "swollen lymph nodes", "bone pain",
        "night sweats", "early satiety", "weight loss", "frequent fractures",
        "nosebleed", "petechiae", "joint pain", "slow growth",
        "dark urine", "itching after bathing", "dizziness", "headache"
    ]
    for i, symptom in enumerate(symptoms, 1):
        print(f"{i}. {symptom}")
    return symptoms

def get_user_symptoms():
    symptoms_list = show_symptoms_menu()
    while True:
        entry = input("\nIntroduce los números de tus síntomas separados por comas (ej: 1,3,5): ").strip()
        if entry == "0":
            return []
        try:
            numbers = [int(n.strip()) for n in entry.split(",")]
            selected = [symptoms_list[i-1] for i in numbers if 1 <= i <= len(symptoms_list)]
            if not selected:
                print("No se seleccionaron síntomas válidos.")
            else:
                return selected
        except Exception:
            print("ERROR: Entrada inválida. Introduce números separados por comas.")

def show_results(facts):
    diagnosis_found = False
    option_counter = 1

    for fact in facts:
        if isinstance(fact, Fact) and hasattr(fact, 'get') and fact.get("disease"):
            if option_counter > 1:
                print(f"\nOPCIÓN {option_counter}:")
            print(f"DIAGNÓSTICO PRELIMINAR: {fact['disease']}")
            print(f"EXPLICACIÓN: {fact['explanation']}")
            print(f"RECOMENDACIÓN: {fact['recommendation']}")
            option_counter += 1
            diagnosis_found = True

    if not diagnosis_found:
        print("\nNo se encontró un diagnóstico específico para los síntomas ingresados.")
        print("Recomendación general: Consulta a un hematólogo para una evaluación detallada.")
```

```python
print(""" 
BIENVENIDO AL SISTEMA EXPERTO DE DIAGNÓSTICO HEMATOLÓGICO
Este sistema identifica posibles enfermedades hematológicas en función de los síntomas.
""")

continuar = True
while continuar:
    symptoms = get_user_symptoms()
    if not symptoms:
        print("No se ingresaron síntomas específicos.")
    else:
        print("\nAnalizando síntomas...")
        facts = get_diagnosis(symptoms)
        show_results(facts)
    answer = input("\n¿Deseas realizar otro diagnóstico? (s/n): ")
    continuar = (SioNo(answer) == "s")

print("\nGracias por usar el sistema.")
```
